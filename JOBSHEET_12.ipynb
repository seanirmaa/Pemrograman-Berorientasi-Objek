```python
# Praktikum 02: Membaca Data CSV dengan Pandas

# Impor library pandas
import pandas as pd

# Nama file CSV yang akan dibaca
NAMA_FILE_CSV = "lokasi_semarang.csv"

def baca_data_lokasi(nama_file: str) -> pd.DataFrame | None:
    """
    Membaca data lokasi dari file CSV menggunakan Pandas.

    Args:
        nama_file (str): Path atau nama file CSV yang akan dibaca.

    Returns:
        pd.DataFrame | None: DataFrame Pandas berisi data jika berhasil,
                             None jika file tidak ditemukan atau error lain.
    """
    print(f"Mencoba membaca file CSV: {nama_file}")
    try:
        # Menggunakan pandas.read_csv untuk membaca file
        # Secara default, read_csv menganggap baris pertama sebagai header
        # dan koma sebagai pemisah (separator).
        dataframe = pd.read_csv(nama_file)
        print(" -> File CSV berhasil dibaca.")
        return dataframe
    except FileNotFoundError:
        # Menangani jika file tidak ditemukan
        print(f" -> ERROR: File '{nama_file}' tidak ditemukan!")
        return None
    except pd.errors.EmptyDataError:
         print(f" -> ERROR: File '{nama_file}' kosong.")
         return None
    except Exception as e:
        # Menangani error umum lainnya saat membaca CSV
        print(f" -> ERROR saat membaca file CSV: {type(e).__name__} - {e}")
        return None

# --- Kode Utama ---
if __name__ == "__main__":
    print("--- Memulai Praktikum 2: Membaca CSV ---")
    # Panggil fungsi untuk membaca data
    df_lokasi = baca_data_lokasi(NAMA_FILE_CSV)

    # Periksa apakah pembacaan berhasil (df_lokasi bukan None)
    if df_lokasi is not None:
        print("\n--- Inspeksi Awal DataFrame ---")

        # 1. Tampilkan 5 baris pertama data menggunakan head()
        print("\n1. Lima Baris Pertama (head()):")
        print(df_lokasi.head())

        # 2. Tampilkan informasi ringkas tentang DataFrame menggunakan info()
        #    (Jumlah baris, nama kolom, tipe data kolom, memori)
        print("\n2. Informasi DataFrame (info()):")
        df_lokasi.info()

        # 3. Tampilkan dimensi DataFrame (jumlah baris, jumlah kolom)
        jumlah_baris, jumlah_kolom = df_lokasi.shape
        print(f"\n3. Dimensi Data:")
        print(f"   Jumlah Lokasi (Baris) : {jumlah_baris}")
        print(f"   Jumlah Atribut (Kolom): {jumlah_kolom}")

        # 4. Tampilkan nama-nama kolom
        print(f"\n4. Nama Kolom:")
        print(list(df_lokasi.columns))
    else:
        print("\nTidak dapat melanjutkan inspeksi karena gagal membaca file CSV.")

    print("\n--- Praktikum 2 Selesai ---")
```

```python
# Praktikum 03: Struktur Kelas OOP (Model Data Geografis)

class Lokasi:
    """Kelas dasar (Parent Class) untuk merepresentasikan koordinat geografis."""
    def __init__(self, nama: str, latitude: float, longitude: float, tipe: str, deskripsi: str):
        self.nama = nama
        self.latitude = float(latitude)
        self.longitude = float(longitude)
        self.tipe = tipe
        self.deskripsi = deskripsi

    def get_koordinat(self) -> tuple:
        """Mengembalikan pasangan latitude dan longitude dalam bentuk tuple."""
        return (self.latitude, self.longitude)

    def tampilkan_info(self) -> str:
        """Mengembalikan string informasi dasar untuk marker."""
        return f"<b>{self.nama}</b> ({self.tipe})<br>{self.deskripsi}"


class TempatWisata(Lokasi):
    """Subclass untuk entitas Wisata, mewarisi dari kelas Lokasi."""
    def __init__(self, nama: str, latitude: float, longitude: float, tipe: str, deskripsi: str):
        super().__init__(nama, latitude, longitude, tipe, deskripsi)

    def tampilkan_info(self) -> str:
        """Polymorphism: Menambahkan ikon visual atau modifikasi informasi tempat wisata."""
        return f"🗺️ <b>[WISATA] {self.nama}</b><br>📌 {self.deskripsi}"


class Kuliner(Lokasi):
    """Subclass untuk entitas Kuliner, mewarisi dari kelas Lokasi."""
    def __init__(self, nama: str, latitude: float, longitude: float, tipe: str, deskripsi: str):
        super().__init__(nama, latitude, longitude, tipe, deskripsi)

    def tampilkan_info(self) -> str:
        """Polymorphism: Menambahkan ikon visual khas makanan/kuliner."""
        return f"🍲 <b>[KULINER] {self.nama}</b><br>😋 {self.deskripsi}"
```

```python
# Praktikum 04: Iterasi DataFrame ke Objek OOP

def buat_objek_dari_dataframe(df: pd.DataFrame) -> list:
    """Mengiterasi baris DataFrame dan mengembalikan daftar objek lokasi."""
    daftar_objek = []

    for index, row in df.iterrows():
        nama = row['Nama']
        lat = row['Latitude']
        lon = row['Longitude']
        tipe = row['Tipe']
        deskripsi = row['Deskripsi']

        # Pengkondisian tipe objek (Polymorphism & Inheritance)
        if "Wisata" in tipe or tipe == "Landmark":
            objek = TempatWisata(nama, lat, lon, tipe, deskripsi)
        elif tipe == "Kuliner":
            objek = Kuliner(nama, lat, lon, tipe, deskripsi)
        else:
            objek = Lokasi(nama, lat, lon, tipe, deskripsi)

        daftar_objek.append(objek)

    return daftar_objek

# Mengubah data dari Praktikum 2 menjadi list objek
if df_lokasi is not None:
    daftar_lokasi_objek = buat_objek_dari_dataframe(df_lokasi)
    print(f"✔ Berhasil membuat {len(daftar_lokasi_objek)} objek berbasis OOP.")

    # Uji coba menampilkan salah satu info objek
    print("\nUji coba pemanggilan metode objek pertama:")
    print(daftar_lokasi_objek[0].tampilkan_info())
```

```python
 # Praktikum 05: Visualisasi Peta Interaktif dengan Folium
import folium

# Inisialisasi peta berpusat di Semarang
koordinat_semarang = [-6.9929, 110.4200]
peta_semarang = folium.Map(location=koordinat_semarang, zoom_start=12)

# Tambahkan marker secara dinamis dari daftar objek lokasi (menggunakan Polymorphism)
for lokasi in daftar_lokasi_objek:
    # Menentukan warna ikon berdasarkan tipe kelas objek
    if isinstance(lokasi, TempatWisata):
        warna_ikon = 'blue'
        ikon_visual = 'cloud'
    elif isinstance(lokasi, Kuliner):
        warna_ikon = 'orange'
        ikon_visual = 'info-sign'
    else:
        warna_ikon = 'green'
        ikon_visual = 'home'

    # Buat marker dan tambahkan ke peta
    folium.Marker(
        location=lokasi.get_koordinat(),
        popup=folium.Popup(lokasi.tampilkan_info(), max_width=300),
        tooltip=f"Klik untuk info {lokasi.nama}",
        icon=folium.Icon(color=warna_ikon, icon=ikon_visual)
    ).add_to(peta_semarang)

# Simpan hasil ke file HTML
peta_semarang.save("peta_semarang.html")
print("✔ Peta berhasil dibuat dan disimpan sebagai 'peta_semarang.html'.")

# Menampilkan peta langsung di Google Colab
peta_semarang
```

```python
# ==========================================
# 1. PERSIAPAN ABSTRACT BASE CLASS & MODEL
# ==========================================
from abc import ABC, abstractmethod
import pandas as pd
import folium
import datetime

class GeoEntity(ABC):
    """
    Abstract Base Class (ABC) sebagai cetak biru (blueprint)
    untuk semua entitas spasial/geografis.
    """
    def __init__(self, nama: str, latitude: float, longitude: float, tipe: str, deskripsi: str):
        self.nama = nama
        self.latitude = float(latitude)
        self.longitude = float(longitude)
        self.tipe = tipe
        self.deskripsi = deskripsi

    @abstractmethod
    def tampilkan_info(self) -> str:
        """Method abstrak yang WAJIB diimplementasikan oleh setiap subclass."""
        pass

    def get_koordinat(self) -> tuple:
        """Mengembalikan pasangan latitude dan longitude dalam bentuk tuple."""
        return (self.latitude, self.longitude)


class TempatWisata(GeoEntity):
    """Subclass untuk entitas Wisata, mewarisi dari GeoEntity."""
    def tampilkan_info(self) -> str:
        return f"🗺️ <b>[WISATA] {self.nama}</b><br>📌 {self.deskripsi}"


class Kuliner(GeoEntity):
    """Subclass untuk entitas Kuliner, mewarisi dari GeoEntity."""
    def tampilkan_info(self) -> str:
        return f"🍲 <b>[KULINER] {self.nama}</b><br>😋 {self.deskripsi}"


class TempatIbadah(GeoEntity):
    """Subclass tambahan untuk Tempat Ibadah, mewarisi dari GeoEntity."""
    def tampilkan_info(self) -> str:
        return f"🕌 <b>[IBADAH] {self.nama}</b><br>✨ {self.deskripsi}"


# ==========================================
# 2. FUNGSI LOAD DATA & EXCEPTION HANDLING
# ==========================================
def load_and_validate_data(nama_file: str) -> list:
    """
    Membaca CSV dengan pengondisian tipe data, validasi kolom,
    dan Exception Handling yang kuat.
    """
    daftar_objek = []
    kolom_wajib = {'Nama', 'Latitude', 'Longitude', 'Tipe', 'Deskripsi'}

    try:
        # Membaca file CSV menggunakan Pandas
        df = pd.read_csv(nama_file)

        # Validasi ketersediaan kolom wajib (KeyError handling secara manual)
        if not kolom_wajib.issubset(df.columns):
            hilang = kolom_wajib - set(df.columns)
            raise KeyError(f"Kolom wajib tidak ditemukan di CSV: {hilang}")

        # Mengiterasi baris demi baris
        for index, row in df.iterrows():
            try:
                nama = str(row['Nama']).strip()
                lat = float(row['Latitude'])    # Memicu ValueError jika bukan angka
                lon = float(row['Longitude'])   # Memicu ValueError jika bukan angka
                tipe = str(row['Tipe']).strip()
                deskripsi = str(row['Deskripsi']).strip()

                # Pengkondisian tipe objek berdasarkan kolom 'Tipe'
                if "Wisata" in tipe or tipe == "Landmark":
                    objek = TempatWisata(nama, lat, lon, tipe, deskripsi)
                elif "Kuliner" in tipe:
                    objek = Kuliner(nama, lat, lon, tipe, deskripsi)
                elif "Ibadah" in tipe:
                    objek = TempatIbadah(nama, lat, lon, tipe, deskripsi)
                else:
                    # Default jika tipe lain tidak masuk spesifikasi khusus
                    objek = TempatWisata(nama, lat, lon, tipe, deskripsi)

                daftar_objek.append(objek)

            except ValueError:
                print(f"⚠ Baris {index+1} dilewati: Koordinat Latitude/Longitude harus berupa angka!")
            except Exception as e:
                print(f"⚠ Gagal memproses baris {index+1}: {e}")

        print(f"✔ Validasi sukses. {len(daftar_objek)} objek berhasil dimuat ke memori.")
        return daftar_objek

    except FileNotFoundError:
        print(f"❌ Error kritis: File '{nama_file}' tidak ditemukan!")
        return []
    except KeyError as ke:
        print(f"❌ Error struktur file: {ke}")
        return []
    except Exception as e:
        print(f"❌ Terjadi kesalahan sistem: {e}")
        return []


# ==========================================
# 3. FUNGSI PENCARIAN / FILTER DATA
# ==========================================
def cari_lokasi(daftar_objek: list, kata_kunci: str) -> list:
    """Memfilter daftar objek berdasarkan nama atau deskripsi (Case-Insensitive)."""
    hasil_pencarian = [
        obj for obj in daftar_objek
        if kata_kunci.lower() in obj.nama.lower() or kata_kunci.lower() in obj.deskripsi.lower()
    ]
    return hasil_pencarian


# ==========================================
# 4. PROSES EKSEKUSI UTAMA (MAIN PROGRAM)
# ==========================================
NAMA_FILE_CSV = "lokasi_semarang.csv"

# Mengisi & memvalidasi objek data
semua_lokasi = load_and_validate_data(NAMA_FILE_CSV)

if semua_lokasi:
    # --- Contoh Fitur Pencarian ---
    kata_kunci_cari = "Lumpia"
    hasil_filter = cari_lokasi(semua_lokasi, kata_kunci_cari)

    print(f"\n--- Hasil Pencarian dengan Kata Kunci '{kata_kunci_cari}' ---")
    for item in hasil_filter:
        print(f"- {item.nama} ({item.tipe})")

    # --- Pembuatan Peta dengan Custom Marker ---
    koordinat_semarang = [-6.9929, 110.4200]
    peta_tugas = folium.Map(location=koordinat_semarang, zoom_start=12)

    for lokasi in semua_lokasi:
        # Penentuan warna marker dan ikon visual berdasarkan tipe objek konkret
        if isinstance(lokasi, TempatWisata):
            warna_ikon = 'blue'
            ikon_visual = 'camera'
        elif isinstance(lokasi, Kuliner):
            warna_ikon = 'orange'
            ikon_visual = 'cutlery'
        elif isinstance(lokasi, TempatIbadah):
            warna_ikon = 'green'
            ikon_visual = 'heart'
        else:
            warna_ikon = 'gray'
            ikon_visual = 'info-sign'

        # Menambahkan marker ke peta
        folium.Marker(
            location=lokasi.get_koordinat(),
            popup=folium.Popup(lokasi.tampilkan_info(), max_width=300),
            tooltip=lokasi.nama,
            icon=folium.Icon(color=warna_ikon, icon=ikon_visual, prefix='fa') # Menggunakan FontAwesome Icon
        ).add_to(peta_tugas)

    # Menyimpan peta ke berkas HTML eksternal
    peta_tugas.save("peta_semarang_tugas.html")
    print("\n✔ Peta tugas interaktif sukses dibuat sebagai 'peta_semarang_tugas.html'.")
```
